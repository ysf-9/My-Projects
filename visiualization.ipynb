```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
year = [2017,2018,2019,2020,2021,2022]
expense = [100000,120000,128000,110000,140000,200000]
profit = [10000,20000,12000,13000,11000,14000]
```

```python
startupX = {
    "year" : year ,
    "expense" : expense ,
    "profit" : profit
}
```

```python
df = pd.DataFrame(startupX)
df.head()
```

```python
df.to_csv("startupX.csv")
```
