```python
ratings = pd.read_csv("/content/Netflix_Dataset_Rating.csv")
ratings.head()
```

```python
movies = pd.read_csv("/content/Netflix_Dataset_Movie.csv")
movies.head()
```

```python
movie_rating = pd.merge(ratings,movies)
movie_rating.head()
```

```python
movie_rating.shape
```

```python
movie_rating = movie_rating.pivot_table(index="User_ID",columns="Name",values='Rating')
movie_rating.head()
```

```python
movies.Name[movies['Name'].str.contains("Wild")]
```

```python
wild_ratings = movie_rating["Wild Things"]
wild_ratings.head()
```

```python
similar_movies = movie_rating.corrwith(wild_ratings)
similar_movies.head()
```

```python
similar_movies.sort_values(ascending=False)
```

```python
average_ratings = movie_rating.mean(axis=0).sort_values(ascending=False)
print("Top 5 movies by average rating:")
display(average_ratings.head())
```
